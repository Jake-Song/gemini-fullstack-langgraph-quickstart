```python
# Simple asyncio examples

import asyncio
import time
import aiohttp
import random
```

```python
# Example 1: Basic async function
async def greet(name, delay):
    """Simple async function that waits and then greets"""
    print(f"Starting to greet {name}...")
    await asyncio.sleep(delay)  # Simulate some work
    print(f"Hello, {name}!")
    return f"Greeted {name}"

# Run a single async function
result = await greet("Alice", 1)
print(f"Result: {result}")
```

```python
# Example 2: Running multiple async functions concurrently
async def fetch_data(url, delay):
    """Simulate fetching data from a URL"""
    print(f"Fetching data from {url}...")
    await asyncio.sleep(delay)
    return f"Data from {url}"

# Run multiple async functions concurrently
start_time = time.time()

# Using asyncio.gather to run multiple coroutines concurrently
results = await asyncio.gather(
    fetch_data("https://api.example.com/users", 2),
    fetch_data("https://api.example.com/posts", 1.5),
    fetch_data("https://api.example.com/comments", 1),
    greet("Bob", 0.5)
)

end_time = time.time()
print(f"\nAll tasks completed in {end_time - start_time:.2f} seconds")
print("Results:", results)
```

```python
# Example 3: Using asyncio.create_task for more control
async def process_item(item_id, processing_time):
    """Process an item with some delay"""
    print(f"Processing item {item_id}...")
    await asyncio.sleep(processing_time)
    return f"Processed item {item_id}"

# Create tasks and manage them individually
tasks = []
for i in range(5):
    delay = random.uniform(0.5, 2.0)  # Random delay between 0.5 and 2 seconds
    task = asyncio.create_task(process_item(i, delay))
    tasks.append(task)

# Wait for all tasks to complete
results = await asyncio.gather(*tasks)
print("All items processed:", results)
```

```python
# Example 4: Async context manager and error handling
async def risky_operation(operation_id, should_fail=False):
    """An operation that might fail"""
    await asyncio.sleep(1)
    if should_fail:
        raise ValueError(f"Operation {operation_id} failed!")
    return f"Operation {operation_id} succeeded"

# Handle errors in async operations
try:
    results = await asyncio.gather(
        risky_operation(1, False),
        risky_operation(2, True),  # This one will fail
        risky_operation(3, False),
        return_exceptions=True  # Don't let one failure stop others
    )
    print("Results with error handling:", results)
except Exception as e:
    print(f"Caught exception: {e}")
```

```python
# Example 5: Real-world example - Making HTTP requests concurrently
async def fetch_url(session, url):
    """Fetch a URL and return the status code"""
    try:
        async with session.get(url) as response:
            return f"{url}: {response.status}"
    except Exception as e:
        return f"{url}: Error - {str(e)}"

async def fetch_multiple_urls():
    """Fetch multiple URLs concurrently"""
    urls = [
        "https://httpbin.org/delay/1",
        "https://httpbin.org/delay/2", 
        "https://httpbin.org/status/200",
        "https://httpbin.org/status/404",
        "https://httpbin.org/json"
    ]
    
    async with aiohttp.ClientSession() as session:
        start_time = time.time()
        
        # Create tasks for all URLs
        tasks = [fetch_url(session, url) for url in urls]
        results = await asyncio.gather(*tasks)
        
        end_time = time.time()
        print(f"Fetched {len(urls)} URLs in {end_time - start_time:.2f} seconds")
        
        for result in results:
            print(result)

# Uncomment the line below to run the HTTP example (requires internet connection)
await fetch_multiple_urls()
```

## Key asyncio Concepts Demonstrated:

1. **`async def`** - Define asynchronous functions
2. **`await`** - Wait for async operations to complete
3. **`asyncio.sleep()`** - Non-blocking delay
4. **`asyncio.gather()`** - Run multiple coroutines concurrently
5. **`asyncio.create_task()`** - Create tasks for more control
6. **Error handling** - Using `return_exceptions=True` in gather
7. **Real-world usage** - HTTP requests with aiohttp

## Benefits of asyncio:
- **Concurrency**: Run multiple I/O operations simultaneously
- **Efficiency**: Don't block the thread while waiting for I/O
- **Scalability**: Handle many concurrent operations with minimal resources

## When to use asyncio:
- I/O bound operations (network requests, file operations, database queries)
- When you need to handle many concurrent operations
- Web scraping, API calls, real-time applications
